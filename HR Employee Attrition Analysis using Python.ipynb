```python
import pandas as pd
df=pd.read_excel(r"C:\Users\Lenovo\Desktop\HR_Data.xlsx")
print(df.head())
print(df.info())
df['Attrition'].value_counts()
df.groupby('Department')['Attrition'].value_counts()
df.groupby('Gender')['Attrition'].value_counts()
df.groupby('Salary')['Attrition'].value_counts()
```

```python
import seaborn as sns
import matplotlib.pyplot as plt
sns.countplot(x='Department',hue='Attrition',data=df)
plt.show()
```

```python
-Sales department has higher attrition -Employees with lower salary leaving more -Employees with lesser experience leaving more -IT department attrition is lower

-Sales department has higher attrition -Employees with low salary leaving more -Employees with less experience leaving more -IT department attrition is lower
```
